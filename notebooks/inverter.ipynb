# Inverter

## Periodic Square Waveform

![Periodic Square Waveform]()

Fourier series expression of a period square waveform (without any delay) with amplitude $V_{dc}$ and time period $T$ is:
$$ V_o = \sum_{n=1,3,5}^{\infty}\frac{4V_{dc}}{n\pi} \sin(nw_ot) $$

where, 
$$ w_o = \frac{2\pi}{T}$$

$$ V_{o,rms} = V_{dc} $$
$$ V_{o1} = \frac{4V_{dc}}{\pi}\sin(w_ot) $$
$$ V_{o1,rms} = \frac{4V_{dc}}{\pi\sqrt{2}} = \frac{2\sqrt{2}}{\pi}V_{dc} $$ 
$$ \text{Distortion Factor, g} = \frac{V_{o1,rms}}{V{o,rms}} = \frac{2\sqrt{2}}{\pi} = 0.9 $$
$$ \text{Total Harmonic Distortion, THD} = \frac{\sqrt{I_{o,rms}^2 - I_{o1,rms}^2}}{I_{o1, rms}} = \sqrt{\frac{1}{g^2} - 1} =  48.43 \% $$

## Single Phase Bridge Voltage Source Inverter

$$ V_o = \sum_{n=1,3,5}^{\infty}\frac{4V_{dc}}{n\pi} \sin(nw_ot) $$

$$ V_{o,rms} = V_{dc} $$
$$ V_{o1} = \frac{4V_{dc}}{\pi}\sin(w_ot) $$
$$ V_{o1,rms} = \frac{4V_{dc}}{\pi\sqrt{2}} = \frac{2\sqrt{2}}{\pi}V_{dc} $$ 
$$ \text{Distortion Factor, g} = \frac{V_{o1,rms}}{V{o,rms}} = \frac{2\sqrt{2}}{\pi} = 0.9 $$
$$ \text{Total Harmonic Distortion, THD} = \frac{\sqrt{I_{o,rms}^2 - I_{o1,rms}^2}}{I_{o1, rms}} = \sqrt{\frac{1}{g^2} - 1} =  48.43 \% $$
