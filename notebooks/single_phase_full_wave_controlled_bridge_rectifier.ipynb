# Single Phase Full Wave Controlled Bridge Rectifier - Continuous Conduction

$$ V_{o,avg} = \frac{2V_m}{\pi} \cos\alpha $$

$$ t_c = \frac{(\pi - \alpha)}{w} $$

$$ I_{T,avg} = \frac{I_o}{2} $$

$$ I_{T,rms} = \frac{I_o}{\sqrt{2}} $$

$$ I_{s,avg} = 0 $$

$$ I_{s,rms} = I_o $$

$$ I_s = \sum_{n=1,3,5}^{\infty} \frac{4I_o}{n\pi}\sin(nw_ot - n\alpha) $$ 

$$ I_{s1, rms} = \frac{4I_o}{\pi \sqrt{2}} = \frac{2\sqrt{2}I_o}{\pi} $$ 

$$ \text{Distortion Factor, g} = \frac{I_{s1}}{I_s} = \frac{2\sqrt{2}}{\pi} = 0.9 $$

$$ \text{THD} = \sqrt{\frac{I_s^2 - I_{s1}^2}{I_{s1}^2}} = \sqrt{\frac{1}{g^2} - 1} = 0.4843 \text{ or } 48.43 \% $$

$$ \text{Fundamental Displacement Factor (FDF)} = \cos\left(\angle V_s - \angle I_{s1}\right) = \cos\alpha$$

$$ \text{Input Power Factor (IPF)} = g * FDF = \frac{I_{s1}}{I_s} \cos\alpha = \frac{2\sqrt{2}}{\pi}\cos\alpha $$
