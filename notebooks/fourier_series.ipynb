# Fourier Series - Application in PWM Inverter

Fourier series of a function $f(w_ot)$ can be given as below:

$$ f(w_ot) = a_o + \sum_{n=1}^\infty a_n \cos(nw_ot) + \sum_{n=1}^\infty b_n \sin(nw_ot)  $$

where:

$$ a_o = \frac{1}{2\pi} \int_0^{2\pi} f(w_ot)d(w_ot) $$

$$ a_n = \frac{2}{2\pi} \int_0^{2\pi} f(w_ot)\cos(nw_ot) d(w_ot) $$

$$ 
b_n =
\frac{2}{2\pi} \int_0^{2\pi} f(w_ot)\sin(nw_ot) d(w_ot)
$$

## Finding $a_o$

$$ f(w_ot) = a_o + \sum_{n=1}^\infty a_n \cos(nw_ot) + \sum_{n=1}^\infty b_n \sin(nw_ot)  $$

$$ \int_0^{2\pi} f(w_ot) d(w_ot) = \int_0^{2\pi} \left(a_o + \sum_{n=1}^\infty a_n \cos(nw_ot) + \sum_{n=1}^\infty b_n \sin(nw_ot)\right) d(w_ot) $$

$$ \int_0^{2\pi} f(w_ot) d(w_ot) = a_o \int_0^{2\pi}  d(w_ot) $$

$$ \int_0^{2\pi} f(w_ot) d(w_ot) = a_o (2\pi - 0) $$

$$ a_o = \frac{1}{2\pi} \int_0^{2\pi} f(w_ot) d(w_ot) $$

Therefore, $a_o$ denotes the average value of $f(w_ot)$.

## Finding $a_n$

$$ f(w_ot) = a_o + \sum_{n=1}^\infty a_n \cos(nw_ot) + \sum_{n=1}^\infty b_n \sin(nw_ot) $$

$$ 
f(w_ot)\cos(nw_ot) = 
a_o \cos(nw_ot) + a_n cos^2(nw_ot) + \cos(nw_ot) \sum_{n \ne n}a_n \cos(nw_ot) + \cos(nw_ot) \sum_{n=1}^\infty b_n \sin(nw_ot)
$$

$$ 
\int_0^{2\pi} f(w_ot)\cos(nw_ot) d(w_ot) = 
\int_0^{2\pi} 
\left(
a_o \cos(nw_ot) + a_n cos^2(nw_ot) + \cos(nw_ot) \sum_{n \ne n}a_n \cos(nw_ot) + \cos(nw_ot) \sum_{n=1}^\infty b_n \sin(nw_ot)
\right)
d(w_ot)
$$

$$ 
\int_0^{2\pi} f(w_ot)\cos(nw_ot) d(w_ot) = 
\int_0^{2\pi} a_n \cos^2(nw_ot) d(w_ot)
$$

$$ 
\int_0^{2\pi} f(w_ot)\cos(nw_ot) d(w_ot) = 
a_n \int_0^{2\pi} \frac{1 + \cos(2nw_ot)}{2} d(w_ot)
$$

$$ 
\int_0^{2\pi} f(w_ot)\cos(nw_ot) d(w_ot) = 
\frac{a_n}{2} \int_0^{2\pi} d(w_ot)
$$

$$ 
\int_0^{2\pi} f(w_ot)\cos(nw_ot) d(w_ot) = 
\frac{a_n}{2} (2\pi - 0)
$$

$$ 
a_n =
\frac{2}{2\pi} \int_0^{2\pi} f(w_ot)\cos(nw_ot) d(w_ot)
$$

## Finding $b_n$

$$ f(w_ot) = a_o + \sum_{n=1}^\infty a_n \cos(nw_ot) + \sum_{n=1}^\infty b_n \sin(nw_ot) $$

$$ 
f(w_ot)\sin(nw_ot) = 
a_o \sin(nw_ot) +
\sin(nw_ot) \sum_{n=1}^\infty a_n \cos(nw_ot) + 
b_n sin^2(nw_ot) + 
\sin(nw_ot) \sum_{n \ne n}b_n \sin(nw_ot)
$$

$$ 
\int_0^{2\pi}
f(w_ot)\sin(nw_ot) = 
\left(
a_o \sin(nw_ot) +
\sin(nw_ot) \sum_{n=1}^\infty a_n \cos(nw_ot) + 
b_n sin^2(nw_ot) + 
\sin(nw_ot) \sum_{n \ne n}b_n \sin(nw_ot)
\right)
d(w_ot)
$$

$$ 
\int_0^{2\pi} f(w_ot)\sin(nw_ot) d(w_ot) = 
\int_0^{2\pi} b_n \sin^2(nw_ot) d(w_ot)
$$

$$ 
\int_0^{2\pi} f(w_ot)\sin(nw_ot) d(w_ot) = 
b_n \int_0^{2\pi} \frac{1 - \cos(2nw_ot)}{2} d(w_ot)
$$

$$ 
\int_0^{2\pi} f(w_ot)\sin(nw_ot) d(w_ot) = 
\frac{b_n}{2} \int_0^{2\pi} d(w_ot)
$$

$$ 
\int_0^{2\pi} f(w_ot)\sin(nw_ot) d(w_ot) = 
\frac{b_n}{2} (2\pi - 0)
$$

$$ 
b_n =
\frac{2}{2\pi} \int_0^{2\pi} f(w_ot)\sin(nw_ot) d(w_ot)
$$

### Case-I: $f(w_ot)$ is neither Even nor Odd

- $a_o \ne 0 $
- $a_n \ne 0 $
- $b_n \ne 0 $

### Case-II: $f(w_ot)$ is Even i.e. $f(w_ot) = f(-w_ot) $
- $a_o = 0 $
- $a_n \ne 0 $
- $b_n = 0 $

### Case-III: $f(w_ot)$ is Odd i.e. $f(w_ot) = -f(-w_ot) $
- $a_o = 0 $
- $a_n = 0 $
- $b_n \ne 0 $

### Case-IV: $f(w_ot)$ is Half-Symmetric

A periodic function $f(w_ot)$ is said to have half wave symmetry, if it satisfies the following condition:

$$ f(w_ot) = -f\left(w_o\left(t \pm \frac{T}{2}\right)\right) = -f(w_ot \pm \pi) $$

Where, $T$ is the time period of the function.

- $ a_o = 0 $
  
- $ 
a_n = 
\begin{cases}
0 & \text{ for even n} \\
\frac{4}{2\pi}\int_0^\pi f(w_ot) \cos(nw_ot) & \text{ for odd n}
\end{cases}
$

- $ 
b_n = 
\begin{cases}
0 & \text{ for even n} \\
\frac{4}{2\pi}\int_0^\pi f(w_ot) \sin(nw_ot) & \text{ for odd n}
\end{cases}
$

### Case-IV: $f(w_ot)$ is Odd & Half-Symmetric

- $ a_o = 0 $
  
- $ a_n = 0$

- $ 
b_n = 
\begin{cases}
0 & \text{ for even n} \\
\frac{4}{2\pi}\int_0^\pi f(w_ot) \sin(nw_ot) & \text{ for odd n}
\end{cases}
$
