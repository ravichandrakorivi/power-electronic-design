## Choosing $C$ Value

$$ \Delta i_L = \Delta i_C $$

$$ I_o = V_o/R $$

$$ i_{L, max} = I_o + \frac{\Delta i_L}{2} $$

$$ i_{L, min} = I_o - \frac{\Delta i_L}{2} $$

$$ \Delta Q = \frac{1}{2} \cdot \frac{T_s}{2} \cdot \frac{\Delta i_L}{2} = \frac{\Delta i_L}{8f_s} $$

But:
$$ \Delta Q = C \Delta V $$

Therefore:
$$ C \Delta V = \frac{\Delta i_L}{8f_s} $$ 

$$ C = \frac{\Delta i_L}{8f_s \Delta V} $$ 

## Choosing $L$ Value

$$ v_L = L \frac{di_L}{dt} $$
When the switch is off:

$$  |-V_o| = L \frac{\Delta i_L}{(1-d)T_s}$$

$$ L = \frac{V_o (1-d)}{f_s \Delta i_L} $$

$$ L_{max} = \frac{V_o (1-d_{min})}{f_s \Delta i_L} $$

But:
$$d_{min} = \frac{V_o}{V_{in, max}} $$

![Boundary of continuous conduction](https://github.com/ravichandrakorivi/power-electronic-design/raw/main/images/buck_converter_continuous_conduction_boundary.jpg)

Therefore, choose $\Delta i_L$ either as:
- $\text{10\% of } I_o $
- $2 I_{o,min}$

- Either increase the load current by reducing load resistance
- Increasing inductance value
